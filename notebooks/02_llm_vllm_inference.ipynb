```python
MODEL_NAME = "Qwen/Qwen2.5-0.5B-Instruct"

def build_prompt(dialogue):
    return f"""You are a helpful assistant. Summarize the following dialogue in one concise paragraph.

Dialogue:
{dialogue}

Summary:"""
```

```python
from vllm import LLM, SamplingParams

llm = LLM(
    model=MODEL_NAME,
    dtype="float16",
    gpu_memory_utilization=0.75,
    max_model_len=2048
)
```

```python
sampling_params = SamplingParams(
    temperature=0.0,
    max_tokens=80
)
```

```python
outputs = llm.generate(prompts, sampling_params)

llm_summaries = []

for output in outputs:
    generated_text = output.outputs[0].text.strip()
    llm_summaries.append(generated_text)

eval_df["qwen_zero_shot_summary"] = llm_summaries
```

```python
for i in range(len(eval_df)):
    row = eval_df.iloc[i]

    print("=" * 100)
    print("TOPIC:", row["topic"])

    print("\nREFERENCE SUMMARY:")
    print(row["summary"])

    print("\nQWEN ZERO-SHOT SUMMARY:")
    print(row["qwen_zero_shot_summary"])
```

```python
rouge_metric = evaluate.load("rouge")
sacrebleu_metric = evaluate.load("sacrebleu")

predictions = eval_df["qwen_zero_shot_summary"].tolist()
references = eval_df["summary"].tolist()

rouge_results = rouge_metric.compute(
    predictions=predictions,
    references=references
)

bleu_results = sacrebleu_metric.compute(
    predictions=predictions,
    references=[[ref] for ref in references]
)

qwen_zero_shot_metrics = {
    "method": "qwen2.5_0.5b_instruct_zero_shot_vllm",
    "eval_size": len(eval_df),
    "bleu": bleu_results["score"],
    "rouge1": rouge_results["rouge1"],
    "rouge2": rouge_results["rouge2"],
    "rougeL": rouge_results["rougeL"],
    "rougeLsum": rouge_results["rougeLsum"],
}

qwen_zero_shot_metrics
```

# New generation summury. Fair metircs

```python
def split_into_sentences(text):
    text = str(text).replace("\n", " ")
    text = re.sub(r"\s+", " ", text).strip()

    sentences = re.split(r"(?<=[.!?])\s+", text)
    sentences = [s.strip() for s in sentences if s.strip()]

    return sentences


def first_last_sentence_baseline(dialogue):
    sentences = split_into_sentences(dialogue)

    if len(sentences) == 0:
        return ""

    if len(sentences) == 1:
        return sentences[0]

    return sentences[0] + " " + sentences[-1]
```

```python
eval_df["first_last_baseline_summary"] = eval_df["dialogue"].apply(first_last_sentence_baseline)
```

```python
baseline_predictions = eval_df["first_last_baseline_summary"].tolist()
references = eval_df["summary"].tolist()

baseline_rouge_results = rouge_metric.compute(
    predictions=baseline_predictions,
    references=references
)

baseline_bleu_results = sacrebleu_metric.compute(
    predictions=baseline_predictions,
    references=[[ref] for ref in references]
)

baseline_eval_metrics = {
    "method": "first_last_sentence_baseline",
    "eval_size": len(eval_df),
    "bleu": baseline_bleu_results["score"],
    "rouge1": baseline_rouge_results["rouge1"],
    "rouge2": baseline_rouge_results["rouge2"],
    "rougeL": baseline_rouge_results["rougeL"],
    "rougeLsum": baseline_rouge_results["rougeLsum"],
}

baseline_eval_metrics
```

```python
comparison_df = pd.DataFrame([
    baseline_eval_metrics,
    qwen_zero_shot_metrics
])

comparison_df
```

```python
comparison_df.to_csv(
    "outputs/metrics/baseline_vs_qwen_zero_shot_metrics.csv",
    index=False
)

eval_df.to_csv(
    "outputs/predictions/baseline_vs_qwen_zero_shot_predictions.csv",
    index=False
)
```
